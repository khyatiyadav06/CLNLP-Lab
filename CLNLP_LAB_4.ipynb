NLTK is used as the NLP toolkit for tokenization and term-frequency analysis.


```python
!pip install nltk -q

import nltk
nltk.download("punkt")
nltk.download("punkt_tab")

from nltk.tokenize import word_tokenize
from collections import Counter
import pandas as pd

# Tokenize and normalize
tokens = word_tokenize(text.lower())

# Keep alphabetic terms
terms = [word for word in tokens if word.isalpha()]

# Calculate frequency
term_frequency = Counter(terms)

# Create DataFrame
tf_df = pd.DataFrame(
    term_frequency.items(),
    columns=["Term", "Frequency"]
).sort_values(
    by="Frequency", ascending=False
).reset_index(drop=True)

print("Term Frequency Table:")
display(tf_df)
```

## Experiment 4.3 — TF, DF, IDF and TF-IDF Without NLP Toolkit

### Question

**Write a Python program without using NLTK, spaCy, or any other NLP toolkit to:**

1. Read multiple text documents.
2. Convert text to lowercase.
3. Remove punctuation.
4. Tokenize the documents.
5. Calculate TF for every term.
6. Calculate DF for every term.
7. Calculate IDF using the logarithmic formula.
8. Calculate TF-IDF for every term in every document.
9. Display Term, TF, DF, IDF, and TF-IDF at each step.
10. Identify the top 10 terms with the highest TF-IDF score for each document.


### Given Input

**Document 1:**  
Natural language processing is a field of artificial intelligence.

**Document 2:**  
Natural language processing helps computers understand human language.

**Document 3:**  
Machine learning is an important part of artificial intelligence.


```python
# Given documents
documents = [
    "Natural language processing is a field of artificial intelligence.",
    "Natural language processing helps computers understand human language.",
    "Machine learning is an important part of artificial intelligence."
]

for i, document in enumerate(documents, 1):
    print(f"Document {i}:")
    print(document)
    print()
```

### 4.3(a) Preprocessing and Tokenization


```python
def preprocess(document):
    # Convert to lowercase
    document = document.lower()

    # Remove punctuation
    document = re.sub(r"[^a-z0-9\s]", " ", document)

    # Tokenize using split()
    tokens = document.split()

    return tokens

tokenized_documents = [preprocess(doc) for doc in documents]

for i, tokens in enumerate(tokenized_documents, 1):
    print(f"Document {i} Tokens:")
    print(tokens)
    print()
```

### 4.3(b) Calculate Term Frequency (TF)

TF is calculated as:

**TF(term, document) = Number of occurrences of term / Total number of terms in the document**


```python
tf_data = []

for doc_id, tokens in enumerate(tokenized_documents, 1):
    counts = Counter(tokens)
    total_terms = len(tokens)

    for term, count in counts.items():
        tf = count / total_terms
        tf_data.append({
            "Document": f"Document {doc_id}",
            "Term": term,
            "TF": tf
        })

tf_df = pd.DataFrame(tf_data)

print("Term Frequency (TF):")
display(tf_df)
```

### 4.3(c) Calculate Document Frequency (DF)

DF represents the number of documents in which a term occurs.


```python
num_documents = len(tokenized_documents)

document_frequency = Counter()

for tokens in tokenized_documents:
    unique_terms = set(tokens)
    for term in unique_terms:
        document_frequency[term] += 1

df_table = pd.DataFrame(
    document_frequency.items(),
    columns=["Term", "DF"]
).sort_values(
    by="Term"
).reset_index(drop=True)

print("Document Frequency (DF):")
display(df_table)
```

### 4.3(d) Calculate Inverse Document Frequency (IDF)

The logarithmic formula used is:

**IDF(term) = log(N / DF(term))**

where **N** is the total number of documents.

### 4.3(e) Calculate TF-IDF

**TF-IDF(term, document) = TF(term, document) × IDF(term)**


```python
tfidf_df = tf_df.merge(
    idf_df[["Term", "DF", "IDF"]],
    on="Term",
    how="left"
)

tfidf_df["TF-IDF"] = tfidf_df["TF"] * tfidf_df["IDF"]

tfidf_df = tfidf_df[
    ["Document", "Term", "TF", "DF", "IDF", "TF-IDF"]
]

print("TF-IDF Table:")
display(tfidf_df.sort_values(["Document", "TF-IDF"], ascending=[True, False]))
```

### 4.3(f) Top 10 Terms with Highest TF-IDF for Each Document


```python
for document in tfidf_df["Document"].unique():
    print("=" * 60)
    print(f"Top 10 TF-IDF Terms — {document}")
    print("=" * 60)

    top_terms = (
        tfidf_df[tfidf_df["Document"] == document]
        .sort_values("TF-IDF", ascending=False)
        .head(10)
    )

    display(top_terms)
```
