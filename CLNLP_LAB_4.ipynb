## 4.1. Perform Term-Frequency Analysis and Named Entity Recognition using a toolkit.

```python
# Install spaCy and download the English model
!pip -q install spacy
!python -m spacy download en_core_web_sm -q

import re
import csv
import spacy
from collections import Counter
from google.colab import files
```

### Upload the input text file

```python
uploaded = files.upload()
filename = next(iter(uploaded))

with open(filename, "r", encoding="utf-8") as f:
    text = f.read()

print("Input file:", filename)
print("\nFirst 500 characters:\n")
print(text[:500])
```

```python
# Calculate term frequency using spaCy tokenization
nlp = spacy.load("en_core_web_sm")
doc = nlp(text.lower())

terms = [token.text for token in doc if token.is_alpha]
frequency = Counter(terms)

# Save frequency output in CSV format
with open("term_frequency_4_1.csv", "w", newline="", encoding="utf-8") as f:
    writer = csv.writer(f)
    writer.writerow(["Term", "Frequency"])
    for term, count in sorted(frequency.items(), key=lambda x: (-x[1], x[0])):
        writer.writerow([term, count])

print("Term frequency CSV created successfully.")
```

```python
# Display top 10 most frequent terms
top10 = frequency.most_common(10)

print("Top 10 Most Frequent Terms")
print("---------------------------")
for term, count in top10:
    print(f"{term}: {count}")
```

```python
# Named Entity Recognition (NER)
print("Named Entities")
print("--------------")

for entity in doc.ents:
    print(f"{entity.text} -> {entity.label_}")
```

## 4.2. Perform Term-Frequency Analysis without using any NLP toolkit.

```python
# Read the same input text file
with open(filename, "r", encoding="utf-8") as f:
    text = f.read()

# Convert to lowercase and remove punctuation
text = text.lower()
text = re.sub(r"[^a-z0-9\s]", " ", text)

# Tokenize manually
words = text.split()

# Calculate frequency using a dictionary
frequency_manual = {}

for word in words:
    if word in frequency_manual:
        frequency_manual[word] += 1
    else:
        frequency_manual[word] = 1

# Save output in CSV format
with open("term_frequency_4_2.csv", "w", newline="", encoding="utf-8") as f:
    writer = csv.writer(f)
    writer.writerow(["Term", "Frequency"])
    for term, count in sorted(frequency_manual.items(), key=lambda x: (-x[1], x[0])):
        writer.writerow([term, count])

print("Term frequency CSV created successfully.")
```

```python
# Display top 10 most frequent terms
top10_manual = sorted(frequency_manual.items(), key=lambda x: (-x[1], x[0]))[:10]

print("Top 10 Most Frequent Terms")
print("---------------------------")
for term, count in top10_manual:
    print(f"{term}: {count}")
```

## 4.3. Calculate TF, DF, IDF and TF-IDF without using any NLP toolkit.

```python
import math
import re

# Given three documents
documents = {
    "Document 1": "Natural language processing is a field of artificial intelligence.",
    "Document 2": "Natural language processing helps computers understand human language.",
    "Document 3": "Machine learning is an important part of artificial intelligence."
}

# Convert to lowercase, remove punctuation and tokenize
tokenized_documents = {}

for name, text in documents.items():
    text = text.lower()
    text = re.sub(r"[^a-z0-9\s]", "", text)
    tokenized_documents[name] = text.split()

print("Tokenized Documents")
print("-------------------")
for name, tokens in tokenized_documents.items():
    print(name, ":", tokens)
```

### Calculate TF for every term

```python
tf = {}

for doc_name, tokens in tokenized_documents.items():
    counts = {}
    for term in tokens:
        counts[term] = counts.get(term, 0) + 1

    total_words = len(tokens)
    tf[doc_name] = {term: count / total_words for term, count in counts.items()}

print("TF Values")
print("---------")
for doc_name, values in tf.items():
    print("\n", doc_name)
    for term, value in sorted(values.items()):
        print(f"{term}: {value:.4f}")
```

### Calculate DF for every term

```python
df = {}

for tokens in tokenized_documents.values():
    for term in set(tokens):
        df[term] = df.get(term, 0) + 1

print("DF Values")
print("---------")
for term, value in sorted(df.items()):
    print(f"{term}: {value}")
```

### Calculate IDF using the logarithmic formula

```python
# IDF = log(N / DF), where N is the total number of documents
N = len(tokenized_documents)
idf = {}

for term, value in df.items():
    idf[term] = math.log(N / value)

print("IDF Values")
print("----------")
for term, value in sorted(idf.items()):
    print(f"{term}: {value:.4f}")
```

### Calculate TF-IDF for every term in every document

```python
tfidf = {}

for doc_name, values in tf.items():
    tfidf[doc_name] = {}
    for term, tf_value in values.items():
        tfidf[doc_name][term] = tf_value * idf[term]

print("TF-IDF Values")
print("-------------")
for doc_name, values in tfidf.items():
    print("\n", doc_name)
    for term, value in sorted(values.items()):
        print(f"{term}: {value:.4f}")
```

### Display Term, TF, DF, IDF and TF-IDF

```python
print(f"{'Document':<12}{'Term':<18}{'TF':<10}{'DF':<8}{'IDF':<10}{'TF-IDF':<10}")
print("-" * 68)

for doc_name, values in tf.items():
    for term, tf_value in sorted(values.items()):
        print(f"{doc_name:<12}{term:<18}{tf_value:<10.4f}{df[term]:<8}{idf[term]:<10.4f}{tfidf[doc_name][term]:<10.4f}")
```

### Identify the top 10 terms with the highest TF-IDF score for each document.

```python
for doc_name, values in tfidf.items():
    top_terms = sorted(values.items(), key=lambda x: (-x[1], x[0]))[:10]

    print(f"\nTop 10 TF-IDF Terms - {doc_name}")
    print("--------------------------------")
    for term, score in top_terms:
        print(f"{term}: {score:.4f}")
```
