## Experiment 3.1 — Porter Stemming

### Question
**Write a Python program to perform stemming using the Porter Stemmer. Apply stemming to the following words and display the original words and their stems:**

`playing, played, plays, studies, studying, connected, connection, computers`

```python
!pip install nltk -q

import nltk
from nltk.stem import PorterStemmer

stemmer = PorterStemmer()
words = ['playing', 'played', 'plays', 'studies', 'studying', 'connected', 'connection', 'computers']

print('Original Word\tStem')
print('-' * 35)
for word in words:
    print(f'{word}\t\t{stemmer.stem(word)}')
```

### Result

The Porter Stemmer successfully reduces the given words to their stems.

## Experiment 3.2 — Lemmatization

### Question
**Write a Python program to perform lemmatization on the following input words:**

`cats, dogs, running, runs, ran, studies, studying, better, children, mice, went, ate, leaves, caring`

```python
import nltk
from nltk.stem import WordNetLemmatizer
from nltk.corpus import wordnet

nltk.download('wordnet')
nltk.download('omw-1.4')

lemmatizer = WordNetLemmatizer()
words = ['cats', 'dogs', 'running', 'runs', 'ran', 'studies', 'studying', 'better', 'children', 'mice', 'went', 'ate', 'leaves', 'caring']

pos_mapping = {
    'cats': wordnet.NOUN, 'dogs': wordnet.NOUN,
    'running': wordnet.VERB, 'runs': wordnet.VERB, 'ran': wordnet.VERB,
    'studies': wordnet.VERB, 'studying': wordnet.VERB,
    'better': wordnet.ADJ, 'children': wordnet.NOUN, 'mice': wordnet.NOUN,
    'went': wordnet.VERB, 'ate': wordnet.VERB,
    'leaves': wordnet.NOUN, 'caring': wordnet.VERB
}

print('Original Word\tLemma')
print('-' * 35)
for word in words:
    print(f'{word}\t\t{lemmatizer.lemmatize(word, pos=pos_mapping[word])}')
```

### Result

The given words are successfully converted to their meaningful lemmas using WordNet Lemmatizer.

## Experiment 3.3 — Regular Expression (Regex)

### Question
**Write a Python program using Regular Expressions (Regex) to extract and process the following information from a text:**

- Email addresses
- URLs
- Mobile numbers
- Hashtags
- Mentions

### Input Text

> Welcome to the Natural Language Processing workshop! For more information, visit https://www.nlpworkshop.com or www.python.org. You can contact the coordinator at nlpworkshop@gmail.com or support@python.org. For registration, call +91-9876543210 or 9123456789. Follow us on social media @NLPWorkshop and @PythonLearner.
```python
import re

text = '''Welcome to the Natural Language Processing workshop! For more information, visit
https://www.nlpworkshop.com or www.python.org. You can contact the coordinator at
nlpworkshop@gmail.com or support@python.org. For registration, call +91-9876543210 or
9123456789. Follow us on social media @NLPWorkshop and @PythonLearner. Share your
experience using #NLP, #Python, and #MachineLearning. You can also visit
https://github.com/NLPWorkshop for the latest updates.'''

print(text)
```

### 1. Extract Email Addresses

```python
email_pattern = r'\b[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}\b'
emails = re.findall(email_pattern, text)

print('Email Addresses:')
for email in emails:
    print(email)
```

### 2. Extract URLs

```python
url_pattern = r'https?://[^\s]+|www\.[^\s]+'
urls = [url.rstrip('.,!?') for url in re.findall(url_pattern, text)]

print('URLs:')
for url in urls:
    print(url)
```

### 3. Extract Mobile Numbers

```python
mobile_pattern = r'(?:\+91[-\s]?)?[6-9]\d{9}'
mobile_numbers = re.findall(mobile_pattern, text)

print('Mobile Numbers:')
for number in mobile_numbers:
    print(number)
```

### 4. Extract Hashtags

```python
hashtag_pattern = r'#[A-Za-z0-9_]+'
hashtags = re.findall(hashtag_pattern, text)

print('Hashtags:')
for hashtag in hashtags:
    print(hashtag)
```

### 5. Extract Mentions

```python
mention_pattern = r'@[A-Za-z0-9_]+'
mentions = re.findall(mention_pattern, text)

print('Mentions:')
for mention in mentions:
    print(mention)
```

## Combined Output

```python
print('=' * 50)
print('REGEX EXTRACTION RESULTS')
print('=' * 50)
print('\nEmail Addresses:', emails)
print('\nURLs:', urls)
print('\nMobile Numbers:', mobile_numbers)
print('\nHashtags:', hashtags)
print('\nMentions:', mentions)
```
